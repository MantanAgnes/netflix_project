# 01 · Lookup array (Workflow 1)
Defines the list of lookup tables and publishes it as a **job task value** (`my_array`).
Downstream `for_each` tasks (bronze Auto Loader, silver) loop over it, so adding a table means adding one entry here.

Each entry: `sourcefolder` = folder to read, `targetfolder` = folder to write.

```python
files = [
    {"sourcefolder": "netflix_directors", "targetfolder": "netflix_directors"},
    {"sourcefolder": "netflix_cast",      "targetfolder": "netflix_cast"},
    {"sourcefolder": "netflix_countries", "targetfolder": "netflix_countries"},
    {"sourcefolder": "netflix_category",  "targetfolder": "netflix_category"},
    {"sourcefolder": "netflix_titles",    "targetfolder": "netflix_titles"}
]
```

```python
# Outside a job this is a no-op, so the notebook is safe to run by hand.
dbutils.jobs.taskValues.set(key="my_array", value=files)
print(f"Published {len(files)} lookup tables as task value 'my_array'")
```
